# 1.3 Práctica: Agente percepción–acción

## Propósito

Implementar un agente sencillo que reciba percepciones del entorno, seleccione acciones y registre una secuencia de percepciones.

El objetivo es relacionar los conceptos de:

- Sensores y percepciones
- Función de agente
- Acciones
- Actuadores
- Secuencia de percepciones
- Medida de desempeño

> Esta práctica utiliza un sistema de control de temperatura como ejemplo didáctico. No pretende representar un sistema industrial real.


## 1. Modelo conceptual

El agente recibe como percepción la temperatura del entorno.

```text
Temperatura
    ↓
 Percepción
    ↓
   Agente
    ↓
Selección de acción
    ↓
Calentar / Enfriar / No actuar
```

Reglas didácticas:

- Si la temperatura es menor a 18 °C → Encender calefacción
- Si la temperatura es mayor a 25 °C → Encender ventilación
- En otro caso → No actuar


````python
def agente_temperatura(temperatura):
    """Selecciona una acción a partir de una percepción de temperatura."""
    if temperatura < 18:
        return "ENCENDER CALEFACCIÓN"
    elif temperatura > 25:
        return "ENCENDER VENTILACIÓN"
    else:
        return "NO ACTUAR"
````

## 2. Probar la función de agente

La función de agente representa, de manera simplificada:

**Percepción → Acción**


````python
temperatura = 27
accion = agente_temperatura(temperatura)

print("Percepción:", temperatura, "°C")
print("Acción:", accion)
````

## 3. Simulación de varias percepciones

La siguiente lista representa una secuencia de percepciones obtenidas en distintos instantes.


````python
percepciones = [16, 18, 21, 24, 27, 29, 23, 17]

for instante, temperatura in enumerate(percepciones, start=1):
    accion = agente_temperatura(temperatura)
    print(f"t{instante}: {temperatura} °C → {accion}")
````

## 4. Registrar la secuencia de percepciones y acciones


````python
historial = []

for instante, temperatura in enumerate(percepciones, start=1):
    accion = agente_temperatura(temperatura)
    historial.append({
        "instante": instante,
        "percepcion": temperatura,
        "accion": accion
    })

historial
````

El historial permite distinguir entre:

- **Percepción actual:** información recibida en un instante
- **Secuencia de percepciones:** historial de información recibida por el agente


## 5. Mostrar el historial de forma ordenada


````python
print("HISTORIAL DEL AGENTE")
print("-" * 50)

for registro in historial:
    print(
        f"t{registro['instante']} | "
        f"Temperatura: {registro['percepcion']} °C | "
        f"Acción: {registro['accion']}"
    )
````

## 8. Ampliación: agente que considera una secuencia reciente

La siguiente versión utiliza las tres últimas percepciones y calcula su promedio.


````python
def agente_con_historial(historial_temperaturas):
    ultimas = historial_temperaturas[-3:]
    promedio = sum(ultimas) / len(ultimas)

    if promedio < 18:
        return "ENCENDER CALEFACCIÓN"
    elif promedio > 25:
        return "ENCENDER VENTILACIÓN"
    else:
        return "NO ACTUAR"

historial_temperaturas = []

for instante, temperatura in enumerate(percepciones, start=1):
    historial_temperaturas.append(temperatura)
    accion = agente_con_historial(historial_temperaturas)

    print(
        f"t{instante}: percepción={temperatura} °C | "
        f"últimas={historial_temperaturas[-3:]} | "
        f"acción={accion}"
    )
````
