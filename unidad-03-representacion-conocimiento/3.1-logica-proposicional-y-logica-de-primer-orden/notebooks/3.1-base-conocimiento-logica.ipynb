### Propósito

El objetivo **no** es volver a trabajar SAT, CNF o DPLL. En este subtema interesa comprender cómo la lógica puede utilizarse como **lenguaje de representación del conocimiento** dentro de un sistema de Inteligencia Artificial.

## 1. Contexto del ejemplo

Trabajaremos con un dominio sencillo de red informática.

Sabemos que:

a) `PC1` es un equipo

b) `PC2` es un equipo

c) `Router1` es un router

d) `PC1` está conectado a `Router1`

e) `PC2` está conectado a `Router1`

f) `Router1` está operativo

Queremos representar además la siguiente regla:

> Si un equipo está conectado a un router operativo, entonces el equipo tiene acceso a la red.

En lógica de primer orden puede expresarse conceptualmente como:

\[
\forall x \forall r \; ((Equipo(x) \land Router(r) \land Conectado(x,r) \land Operativo(r)) \rightarrow TieneAcceso(x))
\]

## 2. Representación de los hechos

En Python utilizaremos conjuntos y tuplas para representar una pequeña base de conocimiento.

Esta implementación es deliberadamente simple. Su propósito es **visualizar los conceptos**, no construir un motor lógico completo.

```python
equipos = {"PC1", "PC2"}
routers = {"Router1"}
operativos = {"Router1"}

conectado = {
    ("PC1", "Router1"),
    ("PC2", "Router1")
}

print("Equipos:", equipos)
print("Routers:", routers)
print("Routers operativos:", operativos)
print("Conexiones:", conectado)
```

### Interpretación

En términos de lógica de primer orden, estos datos representan hechos como:

\[
Equipo(PC1)
\]

\[
Equipo(PC2)
\]

\[
Router(Router1)
\]

\[
Conectado(PC1, Router1)
\]

\[
Operativo(Router1)
\]

La diferencia con una representación puramente proposicional es que aquí podemos distinguir explícitamente **objetos** y **relaciones**.

## 3. Primera consulta sobre la base de conocimiento

Comprobemos si algunos hechos están almacenados directamente.

```python
print("¿PC1 es un equipo?", "PC1" in equipos)
print("¿Router1 es un router?", "Router1" in routers)
print("¿PC1 está conectado a Router1?", ("PC1", "Router1") in conectado)
print("¿Router1 está operativo?", "Router1" in operativos)
```

Observe que hasta este punto solo estamos **consultando hechos explícitos**.

Todavía no hemos derivado conocimiento nuevo.

## 4. Representación de una regla

La regla del dominio es:

> Si `x` es un equipo, `r` es un router, `x` está conectado a `r` y `r` está operativo, entonces `x` tiene acceso a la red.

Crearemos una función que represente esa regla.

```python
def regla_tiene_acceso(equipo, router):
    return (
        equipo in equipos
        and router in routers
        and (equipo, router) in conectado
        and router in operativos
    )

print(
    "¿Se cumplen las condiciones para PC1 y Router1?",
    regla_tiene_acceso("PC1", "Router1")
)
```

La función anterior no representa un sistema de inferencia general. Únicamente implementa **una regla concreta del dominio** para mostrar cómo las premisas pueden conducir a una conclusión.

## 5. Derivación de nuevo conocimiento

Ahora aplicaremos la regla a todos los equipos y routers conocidos.

```python
tiene_acceso = set()

for equipo in equipos:
    for router in routers:
        if regla_tiene_acceso(equipo, router):
            tiene_acceso.add(equipo)

print("Equipos con acceso inferido:", tiene_acceso)
```

### Resultado

La base de conocimiento original no contenía explícitamente:

\[
TieneAcceso(PC1)
\]

ni:

\[
TieneAcceso(PC2)
\]

Sin embargo, ambos hechos fueron **derivados** a partir del conocimiento existente.

Esto representa la idea central:

**Hechos + Reglas → Nuevo conocimiento**

## 6. Consulta de conocimiento derivado

```python
def consulta_tiene_acceso(equipo):
    return equipo in tiene_acceso

print("¿PC1 tiene acceso?", consulta_tiene_acceso("PC1"))
print("¿PC2 tiene acceso?", consulta_tiene_acceso("PC2"))
print("¿PC3 tiene acceso?", consulta_tiene_acceso("PC3"))
```

## 7. ¿Qué ocurre si cambia el conocimiento?

Supongamos ahora que `Router1` deja de estar operativo.

Modificaremos temporalmente la base de conocimiento y volveremos a calcular las conclusiones.

```python
operativos_original = operativos.copy()

operativos.clear()

tiene_acceso_actualizado = set()

for equipo in equipos:
    for router in routers:
        if regla_tiene_acceso(equipo, router):
            tiene_acceso_actualizado.add(equipo)

print("Routers operativos:", operativos)
print("Equipos con acceso inferido:", tiene_acceso_actualizado)

operativos = operativos_original
```

### Reflexión

Las conclusiones dependen del conocimiento disponible.

En este ejemplo sencillo:

**Router operativo → Puede derivarse acceso**

Si eliminamos ese hecho, la misma conclusión deja de obtenerse.

## 8. Comparación conceptual: proposicional vs. primer orden

En lógica proposicional podríamos crear símbolos como:

\[
P = \text{PC1 está conectado a Router1}
\]

\[
R = \text{Router1 está operativo}
\]

\[
A = \text{PC1 tiene acceso}
\]

y expresar:

\[
(P \land R) \rightarrow A
\]

Esto funciona para un caso particular.

En lógica de primer orden podemos representar una regla general:

\[
\forall x \forall r \;
((Conectado(x,r) \land Operativo(r)) \rightarrow TieneAcceso(x))
\]

La segunda formulación puede reutilizarse para múltiples equipos y routers.

## 12. Conclusión

Este notebook muestra una idea fundamental de la IA simbólica:

\[
\text{Conocimiento explícito} + \text{Reglas}
\rightarrow
\text{Conocimiento derivado}
\]

La lógica proposicional permite representar hechos y reglas sencillas, mientras que la lógica de primer orden permite trabajar de manera más natural con objetos, relaciones, variables y reglas generales.
