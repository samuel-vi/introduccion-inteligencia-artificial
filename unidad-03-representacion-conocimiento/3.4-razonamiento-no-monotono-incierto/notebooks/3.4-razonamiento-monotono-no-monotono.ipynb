# 3.4 Razonamiento monotónico y no monotónico

**Asignatura:** Introducción a la Inteligencia Artificial  
**Unidad 3:** Representación del conocimiento y razonamiento

## Propósito

Comparar de manera práctica el comportamiento de un sistema de razonamiento **monotónico** y uno **no monotónico**, observando qué ocurre cuando aparece nueva información.

El objetivo principal es comprender que:

a) En el razonamiento monotónico, las conclusiones previas se conservan

b) En el razonamiento no monotónico, una conclusión puede revisarse o retirarse

c) La ausencia de información no siempre significa falsedad

d) Las reglas por defecto y las excepciones permiten representar conocimiento revisable


## 1. Ejemplo base: aves y pingüinos

Trabajaremos con el ejemplo clásico:

```text
Ave(Tweety)
```

Regla general:

```text
Normalmente:
Ave(X) → PuedeVolar(X)
```

Posteriormente aparece nueva información:

```text
Pinguino(Tweety)
```

y sabemos que los pingüinos constituyen una excepción.


## 2. Razonamiento monotónico

En este primer ejemplo, una vez que una conclusión es derivada, se conserva aunque se agregue nueva información.


````python
hechos_monotonicos = {
    "Ave(Tweety)"
}

conclusiones_monotonicas = set()

if "Ave(Tweety)" in hechos_monotonicos:
    conclusiones_monotonicas.add("PuedeVolar(Tweety)")

print("Hechos iniciales:")
for h in sorted(hechos_monotonicos):
    print("-", h)

print("\nConclusiones:")
for c in sorted(conclusiones_monotonicas):
    print("-", c)
````

Ahora agregamos nueva información:

```text
Pinguino(Tweety)
```

En este ejemplo monotónico sencillo, la conclusión previa no se elimina automáticamente.


````python
hechos_monotonicos.add("Pinguino(Tweety)")

print("Hechos después de agregar nueva información:")
for h in sorted(hechos_monotonicos):
    print("-", h)

print("\nConclusiones que se conservan:")
for c in sorted(conclusiones_monotonicas):
    print("-", c)
````

### Observación

Aunque ahora sabemos que Tweety es un pingüino, el sistema monotónico mantiene:

```text
PuedeVolar(Tweety)
```

La nueva información **agrega conocimiento**, pero no elimina automáticamente lo ya concluido.


## 3. Razonamiento no monotónico

Ahora construiremos una versión sencilla donde la conclusión puede revisarse cuando aparece una excepción.


````python
def razonar_no_monotonico(hechos):
    conclusiones = set()

    # Regla por defecto:
    # normalmente, si es ave y no sabemos que es pingüino, asumimos que puede volar
    if "Ave(Tweety)" in hechos and "Pinguino(Tweety)" not in hechos:
        conclusiones.add("PuedeVolar(Tweety)")

    # Regla específica:
    # si es pingüino, entonces no puede volar
    if "Pinguino(Tweety)" in hechos:
        conclusiones.discard("PuedeVolar(Tweety)")
        conclusiones.add("NoPuedeVolar(Tweety)")

    return conclusiones
````

````python
hechos_no_monotonicos = {
    "Ave(Tweety)"
}

conclusiones_1 = razonar_no_monotonico(
    hechos_no_monotonicos
)

print("Estado inicial:")
print("Hechos:", hechos_no_monotonicos)
print("Conclusiones:", conclusiones_1)
````

Agregamos ahora la excepción:


````python
hechos_no_monotonicos.add("Pinguino(Tweety)")

conclusiones_2 = razonar_no_monotonico(
    hechos_no_monotonicos
)

print("Después de agregar nueva información:")
print("Hechos:", hechos_no_monotonicos)
print("Conclusiones:", conclusiones_2)
````

### Comparación

Antes:

```text
Ave(Tweety)
→ PuedeVolar(Tweety)
```

Después:

```text
Ave(Tweety)
Pinguino(Tweety)
→ NoPuedeVolar(Tweety)
```

La conclusión anterior fue revisada.


## 4. Visualización del cambio

Compararemos el estado antes y después de la nueva información.


````python
antes = {"PuedeVolar(Tweety)"}
despues = {"NoPuedeVolar(Tweety)"}

retiradas = antes - despues
nuevas = despues - antes

print("Conclusiones retiradas:")
for x in retiradas:
    print("-", x)

print("\nNuevas conclusiones:")
for x in nuevas:
    print("-", x)
````

## 5. Ejemplo adicional: acceso de un estudiante

Regla general:

```text
Si un estudiante está inscrito y tiene cuenta activa,
normalmente puede acceder.
```

Excepción:

```text
Si la cuenta está bloqueada,
no puede acceder.
```


````python
def acceso_estudiante(hechos):
    conclusiones = set()

    if (
        "Inscrito(Ana)" in hechos
        and "CuentaActiva(Ana)" in hechos
        and "CuentaBloqueada(Ana)" not in hechos
    ):
        conclusiones.add("PuedeAcceder(Ana)")

    if "CuentaBloqueada(Ana)" in hechos:
        conclusiones.discard("PuedeAcceder(Ana)")
        conclusiones.add("NoPuedeAcceder(Ana)")

    return conclusiones
````

````python
hechos_acceso = {
    "Inscrito(Ana)",
    "CuentaActiva(Ana)"
}

print("Estado inicial:")
print(acceso_estudiante(hechos_acceso))

hechos_acceso.add("CuentaBloqueada(Ana)")

print("\nCon nueva información:")
print(acceso_estudiante(hechos_acceso))
````

## 6. Ejemplo adicional: sistema de riego

Regla por defecto:

```text
Si el suelo está seco, normalmente se activa el riego.
```

Excepción:

```text
Si existe lluvia inminente, no se activa el riego.
```


````python
def decision_riego(hechos):
    conclusiones = set()

    if "SueloSeco" in hechos and "LluviaInminente" not in hechos:
        conclusiones.add("ActivarRiego")

    if "LluviaInminente" in hechos:
        conclusiones.discard("ActivarRiego")
        conclusiones.add("NoActivarRiego")

    return conclusiones
````

````python
hechos_riego = {
    "SueloSeco"
}

print("Estado inicial:")
print(decision_riego(hechos_riego))

hechos_riego.add("LluviaInminente")

print("\nCon nueva información:")
print(decision_riego(hechos_riego))
````

## 7. Verdadero, falso y desconocido

En muchos problemas, la ausencia de un hecho no significa que sea falso.

Consideremos:

```text
Inscrito(Ana)
Inscrito(Luis)
```

y preguntamos:

```text
¿Inscrito(Carlos)?
```


````python
hechos_estudiantes = {
    "Inscrito(Ana)",
    "Inscrito(Luis)"
}

consulta = "Inscrito(Carlos)"

if consulta in hechos_estudiantes:
    estado = "verdadero"
else:
    estado = "desconocido"

print(f"{consulta}: {estado}")
````

En un supuesto de **mundo abierto**, el resultado correcto es:

```text
desconocido
```

No debemos concluir automáticamente:

```text
NoInscrito(Carlos)
```


## 8. Mundo cerrado

En un supuesto de mundo cerrado, la ausencia del hecho puede interpretarse como falsedad.


````python
def consultar_mundo_cerrado(hecho, base):
    return "verdadero" if hecho in base else "falso"

print(
    "Inscrito(Carlos):",
    consultar_mundo_cerrado(
        "Inscrito(Carlos)",
        hechos_estudiantes
    )
)
````

## 9. Mundo abierto

En un supuesto de mundo abierto, lo no conocido se mantiene como desconocido.


````python
def consultar_mundo_abierto(hecho_positivo, hecho_negativo, base):
    if hecho_positivo in base:
        return "verdadero"
    if hecho_negativo in base:
        return "falso"
    return "desconocido"

print(
    "Inscrito(Carlos):",
    consultar_mundo_abierto(
        "Inscrito(Carlos)",
        "NoInscrito(Carlos)",
        hechos_estudiantes
    )
)
````
